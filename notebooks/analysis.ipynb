### Step 1: Import Core Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

print("Libraries successfully imported!")
```

### Step 2: Load Dataset and Initial Inspection

```python
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

### Step 3: Data Quality & Overall Attrition Rate

```python
print(f"Total Missing Values: {df.isnull().sum().sum()}")
attrition_counts = df['Attrition'].value_counts()
attrition_pct = df['Attrition'].value_counts(normalize=True) * 100
pd.DataFrame({'Count': attrition_counts, 'Percentage (%)': attrition_pct.round(2)})
```

### Step 4: Business Question 1 — Does OverTime drive turnover?

```python
ot_summary = df.groupby('OverTime')['Attrition'].value_counts(normalize=True).unstack() * 100
print(ot_summary.round(2))

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='OverTime', hue='Attrition', palette={'Yes': '#e53935', 'No': '#1e88e5'})
plt.title("Attrition Count by OverTime Status", fontsize=14, fontweight='bold')
plt.xlabel("Works OverTime")
plt.ylabel("Employee Count")
plt.show()
```

### Step 5: Business Question 2 — Which Department has the highest attrition?

```python
dept_summary = df.groupby('Department')['Attrition'].value_counts(normalize=True).unstack() * 100
print(dept_summary.round(2))

plt.figure(figsize=(9, 4))
sns.countplot(data=df, x='Department', hue='Attrition', palette={'Yes': '#e53935', 'No': '#1e88e5'})
plt.title("Attrition by Department", fontsize=14, fontweight='bold')
plt.ylabel("Employee Count")
plt.show()
```

### Step 6: Business Question 3 — Monthly Income Comparison

```python
income_summary = df.groupby('Attrition')['MonthlyIncome'].agg(['mean', 'median', 'std']).round(2)
print(income_summary)

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome', palette={'Yes': '#e53935', 'No': '#1e88e5'})
plt.title("Monthly Income Distribution by Attrition", fontsize=14, fontweight='bold')
plt.xlabel("Attrition")
plt.ylabel("Monthly Income ($)")
plt.show()
```
